```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Sample Dataset Load karna
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
df = pd.read_csv(url)

print("--- Dataset ke pehle 5 rows ---")
print(df.head())

# 2. Data Check karna (Rows, columns, aur missing values)
print("\n--- Data ki jankari ---")
print(df.info())

# 3. Data Cleaning: Duplicate rows hatana aur missing values theek karna
df.drop_duplicates(inplace=True)
df.fillna(method='ffill', inplace=True)
print("\nData clean ho chuka hai!")
# 4. Data Visualization (Histogram aur Line Chart banana)
plt.figure(figsize=(8, 5))
plt.hist(df['sepal_length'], bins=10, color='skyblue', edgecolor='black')
plt.title('Sepal Length Distribution (Histogram)')
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(df['sepal_width'], color='orange', marker='o')
plt.title('Sepal Width Trend (Line Chart)')
plt.xlabel('Index')
plt.ylabel('Sepal Width')
plt.show()
```
